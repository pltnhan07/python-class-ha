OS và OS.Path trong Python là hai công cụ quan trọng giúp thao tác với hệ thống file và folder một cách linh hoạt. Nhờ công cụ này, lập trình viên có thể quản lý đường dẫn, kiểm tra sự tồn tại của file, thực hiện các thao tác khác một cách đơn giản. Trong bài viết này, mình sẽ giới thiệu chi tiết về OS và OS.Path trong Python, cách sử dụng chúng hiệu quả và những tình huống áp dụng thực tế.

### **1. Module OS**
Module OS trong Python cung cấp nhiều phương thức hữu ích để quản lý file và folder, giúp tương tác với hệ điều hành dễ dàng hơn. Dưới đây là một số phương thức quan trọng:

| STT | Phương thức | Mô tả |
| :---: | :--- | :--- |
| 1 | `os.access(path, mode)` | Kiểm tra quyền truy cập vào path dựa trên mode. |
| 2 | `os.chdir(path)` | Thay đổi folder làm việc hiện tại thành path. |
| 3 | `os.chmod(path, mode)` | Thay đổi quyền truy cập của path thành mode. |
| 4 | `os.getcwd()` | Lấy folder làm việc hiện tại. |
| 5 | `os.listdir(path)` | Trả về danh sách các file và folder trong path. |
| 6 | `os.mkdir(path[, mode])` | Tạo một folder mới tại path. |
| 7 | `os.makedirs(path[, mode])` | Tạo folder cùng với các folder cha nếu chưa tồn tại. |
| 8 | `os.remove(path)` | Xóa file tại path. |
| 9 | `os.rmdir(path)` | Xóa folder rỗng tại path. |
| 10 | `os.rename(src, dst)` | Đổi tên file/folder từ src sang dst. |
| 11 | `os.stat(path)` | Lấy thông tin trạng thái của path. |
| 12 | `os.walk(top[, topdown=True[, onerror=None[, followlinks=False]]])` | Duyệt qua cây folder, tạo danh sách các folder và file. |

### **2. Module OS.Path**
Module OS.Path trong Python cung cấp một loạt các phương thức hữu ích để thao tác với file và folder. Dưới đây là danh sách các phương thức phổ biến:

| STT | Phương thức | Mô tả |
| :---: | :--- | :--- |
| 1 | `os.path.abspath(path)` | Trả về đường dẫn tuyệt đối của một tập tin hoặc folder. |
| 2 | `os.path.basename(path)` | Trả về tên tập tin từ đường dẫn đầy đủ. |
| 3 | `os.path.commonprefix(list)` | Trả về tiền tố chung dài nhất của các đường dẫn trong danh sách. |
| 4 | `os.path.dirname(path)` | Trả về folder chứa tập tin. |
| 5 | `os.path.exists(path)` | Kiểm tra xem đường dẫn có tồn tại không. |
| 6 | `os.path.getsize(path)` | Lấy kích thước tập tin tính bằng byte. |
| 7 | `os.path.isfile(path)` | Kiểm tra xem đường dẫn có phải là tập tin không. |
| 8 | `os.path.isdir(path)` | Kiểm tra xem đường dẫn có phải là folder không. |
| 9 | `os.path.join(path1, path2, ...)` | Nối các thành phần đường dẫn một cách thông minh. |
| 10 | `os.path.realpath(path)` | Trả về đường dẫn tuyệt đối, loại bỏ các liên kết tượng trưng (symlink). |
| 11 | `os.path.splitext(path)` | Tách phần mở rộng của tập tin. |
| 12 | `os.path.split(path)` | Chia đường dẫn thành folder và tên tập tin. |

**Ví dụ**: Tạo folder mới

```python
import os

folder_name = "thu_muc_moi"

# Kiểm tra để tránh lỗi FileExistsError nếu folder đã tồn tại
if not os.path.exists(folder_name):
    os.mkdir(folder_name)
    print(f"Đã tạo folder: '{folder_name}'")
else:
    print(f"Folder '{folder_name}' đã tồn tại.")
```

**Ví dụ**: Tạo cấu trúc folder nhiều tầng

Khi cần tạo một đường dẫn dài mà các folder cha bên ngoài chưa hề tồn tại. Nó sẽ tự động tạo toàn bộ chuỗi folder đó.

```python
import os

nested_path = "class/2026/python"

# Tham số exist_ok=True giúp Python không báo lỗi nếu chuỗi thư mục này đã có sẵn
os.makedirs(nested_path, exist_ok=True)
print(f"Đã tạo thành công chuỗi folder: '{nested_path}'")
```

**Ví dụ**: Nối các thành phần đường dẫn

Nối các chuỗi thành một đường dẫn hoàn chỉnh. Nó sẽ tự động sử dụng đúng dấu gạch chéo tùy theo hệ điều hành (dùng \ cho Windows, và / cho Mac/Linux), giúp code không bị lỗi khi chạy trên các máy khác nhau.

```python
import os

folder = "2026/python"
sub_folder = "images"
file_name = "img.jpg"

# Nối các phần lại với nhau
full_path = os.path.join(folder, sub_folder, file_name)

print(full_path) 
```

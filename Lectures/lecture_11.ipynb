# **Giải thích vi phân (Differentiation)**

### **1. Giải tích vi phân cơ sở**
- Vạn vật luôn chuyển động. 
- Đại số hay hình học cơ bản chỉ giúp đo lường thế giới ở những trạng thái tĩnh (ví dụ: diện tích của một mảnh đất, thể tích của một khối hộp). 
- Khi các đại lượng bắt đầu vận động và biến đổi (xe tăng tốc, dân số tăng trưởng,...), toán học tĩnh trở nên bất lực.

=> Giải tích vi phân (Differentiation) là để giải quyết bài toán này: Đo lường sự thay đổi. 
- Nhiệm vụ của vi phân: tính toán chính xác tốc độ biến thiên tức thời của một hàm số tại bất kỳ thời điểm hay vị trí nào.

#### **1.1. Đạo hàm và Giới hạn**
**Định nghĩa**:

Cho một hàm số $f(x)$ liên tục trên một khoảng.
- Đạo hàm (Derivative): Ký hiệu là $f'(x)$ hoặc $\frac{df}{dx}$, đo lường tỷ lệ thay đổi của hàm số $f$ theo biến $x$. 
    - Về mặt hình học, đây là hệ số góc của đường tiếp tuyến. 
    - Về mặt đại số, nó được định nghĩa nghiêm ngặt bằng giới hạn (limit) của tỷ số biến thiên khi sự thay đổi của đầu vào $\Delta x$ tiến dần về 0:
    $$f'(x_0) = \lim_{\Delta x \to 0} \frac{f(x_0 + \Delta x) - f(x_0)}{\Delta x}$$
- Vi phân (Differential): Ký hiệu là $df$, biểu diễn một lượng thay đổi tuyến tính cực nhỏ (vô cùng bé) của giá trị hàm số, được tính bằng tích của đạo hàm và lượng thay đổi cực nhỏ của biến độc lập $dx$:
$$df = f'(x)dx$$

=> Nếu đạo hàm là "tốc độ" tại một thời điểm, thì vi phân chính là "bước dịch chuyển" thực tế cực nhỏ sinh ra từ tốc độ đó.

**Ví dụ 1**: Động học trong Vật lý (Vận tốc và Gia tốc)
- Hãy tưởng tượng bạn đang lái xe từ TP.HCM đi Đà Lạt. Quãng đường đi được biểu diễn bằng một hàm số theo thời gian $s(t)$.
- Nếu bạn chỉ lấy tổng quãng đường chia cho tổng thời gian, bạn có một con số vận tốc trung bình chung chung. Nhưng giải tích vi phân cho phép bạn biết chính xác tốc độ của xe ngay tại giây thứ 10. 
- Đạo hàm bậc 1 của quãng đường chính là vận tốc tức thời $v(t) = s'(t)$ (đây chính là con số đang nhảy trên đồng hồ công-tơ-mét của xe). 
- Tiếp tục, đạo hàm bậc 2 chính là gia tốc $a(t) = v'(t) = s''(t)$, cho biết bạn đang đạp ga tăng tốc hay đạp phanh giảm tốc gắt đến mức nào.

**Ví dụ 2**: Dò biên vật thể trong Xử lý ảnh
- Một bức ảnh kỹ thuật số thực chất là một ma trận điểm ảnh rời rạc $I(x, y)$, với $I$ là mức độ sáng.
- Giả sử có một bức ảnh hình một chú chó, làm sao máy tính biết được đâu là mép tai của chú chó, đâu là bức tường nền ở phía sau? Thuật toán nhận diện rằng: Cạnh của vật thể luôn là nơi cường độ ánh sáng thay đổi đột biến nhất (ví dụ: chuyển gắt từ lông chó màu đen sang nền tường màu trắng). 
- Bằng cách tính đạo hàm không gian (vi phân của ma trận ảnh theo chiều ngang $x$ và dọc $y$), những vùng có giá trị đạo hàm cao sẽ sáng lên thành đường biên, trong khi những vùng màu trơn (đạo hàm bằng 0 do không có sự thay đổi ánh sáng) sẽ bị triệt tiêu thành nền đen. Đây chính là cốt lõi của toán tử Sobel dùng để trích xuất đặc trưng ảnh.

#### **1.2. Các đạo hàm cơ bản**
Trong thực tế, chúng ta hiếm khi phải dùng giới hạn $\lim$ để tính đạo hàm từ con số không. Các nhà toán học đã chứng minh sẵn đạo hàm của các hàm số thông dụng.
| Phân loại | Hàm số $f(x)$ | Đạo hàm $f'(x)$ |
| :--- | :--- | :--- |
| **Hằng số** | $c$ | $0$ |
| **Đa thức** | $x^a$ | $a x^{a-1}$ |
| **Hàm mũ** | $e^x$ | $e^x$ |
| | $a^x$ | $a^x \ln(a)$ |
| **Hàm Logarit** | $\ln(x)$ | $\frac{1}{x}$ |
| **Lượng giác** | $\sin(x)$ | $\cos(x)$ |
| | $\cos(x)$ | $-\sin(x)$ |
| | $\tan(x)$ | $1 + \tan^2(x)$ hoặc $\frac{1}{\cos^2(x)}$ |
| **Lượng giác ngược** | $\arcsin(x)$ | $\frac{1}{\sqrt{1-x^2}}$ |
| | $\arccos(x)$ | $\frac{-1}{\sqrt{1-x^2}}$ |
| | $\arctan(x)$ | $\frac{1}{1+x^2}$ |

#### **1.3. Các quy tắc đạo hàm**
Khi các hàm cơ bản độc lập kết hợp lại với nhau tạo thành một phương trình phức tạp, ta không thể lấy đạo hàm một cách tùy tiện mà phải tuân theo 4 quy tắc vận hành chính yếu sau:

Cho $f, g$ là các hàm số theo $x$ (f(x) và g(x)):
- Quy tắc Tổng/Hiệu (Sum/Difference Rule): Đạo hàm của một tổng bằng tổng các đạo hàm.
    $$(f \pm g)' = f' \pm g'$$
- Quy tắc Tích (Product Rule): Đạo hàm của tích hai hàm số.
    $$(f \cdot g)' = f'g + fg'$$
- Quy tắc Thương (Quotient Rule): Đạo hàm của phép chia hai hàm số (hay còn gọi là đạo hàm của phân số).
    $$\left(\frac{f}{g}\right)' = \frac{f'g - fg'}{g^2}$$
- Quy tắc Chuỗi (Đạo hàm hàm hợp): Khi một hàm số được lồng ghép bên trong một hàm số khác (ví dụ $y = f(u)$ và $u = g(x)$), đạo hàm của toàn bộ hệ thống được tính bằng cách nhân chuỗi các đạo hàm thành phần từ ngoài vào trong:
    $$(f(g(x)))' = f'(g(x)) \cdot g'(x)$$
    - Ví dụ: Tính đạo hàm của $y = \sin(x^2)$.
        - Hàm bên ngoài là $\sin(u)$
        - Hàm bên trong là $u = x^2$.
        - Áp dụng Chain Rule: $y' = \sin(x^2)' \cdot (x^2)' = \cos(x^2) \cdot 2x = 2x\cos(x^2)$.

#### **1.4. Khai triển Taylor**
**Bối cảnh**:
- Trong toán học giải tích và các ngành khoa học ứng dụng, chúng ta thường xuyên phải làm việc với các hàm siêu việt (transcendental functions) mang tính chất phi tuyến phức tạp như $\sin(x)$, $\cos(x)$, $e^x$, hay $\ln(x)$. Tuy nhiên, các hệ thống tính toán (từ bộ vi xử lý máy tính cho đến các thuật toán tối ưu hóa) không thể trực tiếp xử lý các hàm này một cách nguyên bản. Chúng chỉ được thiết kế để thực hiện các phép toán số học cơ bản: cộng, trừ, nhân và chia.
- Khai triển Taylor ra đời nhằm giải quyết rào cản này. Về bản chất, nó cung cấp một phương pháp hệ thống để xấp xỉ (approximate) các hàm số phức tạp bằng các đa thức (polynomials). Đa thức là dạng hàm số lý tưởng nhất trong tính toán vì chúng dễ lấy đạo hàm, dễ lấy tích phân và có thể tính toán xấp xỉ chính xác bằng các phép cộng và nhân lũy thừa. Nhờ Khai triển Taylor, ta có thể chuyển đổi bài toán từ không gian phi tuyến phức tạp về không gian đa thức dễ kiểm soát hơn tại lân cận của một điểm xác định.

**Định nghĩa**:
- Khai triển Taylor của một hàm số $f(x)$ khả vi vô hạn tại lân cận một điểm $x_0$ (gọi là điểm khai triển hay điểm neo) là một chuỗi lũy thừa (power series) vô hạn. Chuỗi này được xây dựng hoàn toàn dựa trên giá trị của hàm số và các đạo hàm bậc cao của nó tại chính điểm $x_0$ đó.
- Khi điểm khai triển được chọn là gốc tọa độ $x_0 = 0$, Khai triển Taylor được gọi bằng một tên định danh cụ thể là Khai triển Maclaurin.

*Hàm số khả vi là hàm số có đạo hàm tại mọi điểm thuộc miền xác định của nó.*

**Công thức**:
- Công thức Khai triển Taylor tổng quát của hàm $f(x)$ tại lân cận điểm $x_0$ được biểu diễn như sau:
    $$f(x) = \sum_{n=0}^{\infty} \frac{f^{(n)}(x_0)}{n!}(x - x_0)^n$$
- Khai triển tường minh chuỗi này, ta có:
    $$f(x) = f(x_0) + \frac{f'(x_0)}{1!}(x - x_0) + \frac{f''(x_0)}{2!}(x - x_0)^2 + \frac{f'''(x_0)}{3!}(x - x_0)^3 + \dots$$

trong đó:
- $f(x_0)$: Thành phần bậc 0 (hằng số). Đảm bảo chuỗi xấp xỉ luôn đi qua đúng điểm tọa độ $(x_0, f(x_0))$ của hàm số gốc.
- $f^{(n)}(x_0)$: Đạo hàm bậc $n$ tại điểm $x_0$. Đại lượng này mang thông tin về mặt hình học của hàm số tại điểm đó (bậc 1 kiểm soát độ dốc, bậc 2 kiểm soát độ cong/Hessian, các bậc cao hơn kiểm soát các điểm uốn và sự biến thiên phức tạp).
- $(x - x_0)^n$: Cơ sở đa thức. Khoảng cách $(x - x_0)$ thể hiện mức độ sai lệch khi biến $x$ rời xa điểm neo $x_0$. Bậc $n$ càng cao, thành phần này càng chi phối độ uốn của xấp xỉ ở xa $x_0$.
- $n!$ (Giai thừa): Đóng vai trò là hệ số chuẩn hóa (normalization factor). Nó giúp làm giảm trọng số của các đa thức bậc cao, đảm bảo chuỗi hội tụ và không bị bùng nổ giá trị khi $x$ biến thiên.

**Ví dụ**: Áp dụng định luật II Newton cho một con lắc đơn có chiều dài sợi dây là $L$ và gia tốc trọng trường $g$, phương trình vi phân mô tả chuyển động góc $\theta(t)$ là:
    $$\frac{d^2\theta}{dt^2} + \frac{g}{L}\sin(\theta) = 0$$
Để đơn giản hóa phương trình trên, các nhà vật lý đưa ra giả định dao động nhỏ (small-angle approximation), tức là góc $\theta$ lân cận giá trị 0. Thực hiện Khai triển Maclaurin (Taylor tại mốc $\theta_0 = 0$) cho hàm $f(\theta) = \sin(\theta)$.

Tính các đạo hàm tại $\theta_0 = 0$:
- $f(0) = \sin(0) = 0$
- $f'(0) = \cos(0) = 1$
- $f''(0) = -\sin(0) = 0$
- $f'''(0) = -\cos(0) = -1$
- $f^{(4)}(0) = \sin(0) = 0$

Thay vào công thức tổng quát:
    $$\sin(\theta) = 0 + \frac{1}{1!}\theta + \frac{0}{2!}\theta^2 + \frac{-1}{3!}\theta^3 + \frac{0}{4!}\theta^4 + \dots$$
    $$\sin(\theta) = \theta - \frac{\theta^3}{6} + \frac{\theta^5}{120} - \dots$$

Vì góc $\theta$ rất nhỏ (thường xét $\theta \ll 1$ radian), các giá trị bậc cao như $\theta^3$ và $\theta^5$ tiến rất nhanh về 0 và trở nên không đáng kể. Do đó, ta có thể ngắt chuỗi Taylor ở bậc 1:
    $$\sin(\theta) \approx \theta$$
    
Thay xấp xỉ tuyến tính này vào phương trình vi phân ban đầu, ta thu được:
    $$\frac{d^2\theta}{dt^2} + \frac{g}{L}\theta = 0$$

### **2. Đạo hàm trong không gian đa biến**

#### **2.1. Đạo hàm riêng (partial derivatives)**
Đối với một hàm số đa biến $f: \mathbb{R}^n \to \mathbb{R}$, nhận đầu vào là một véc-tơ $\mathbf{x} = [x_1, x_2, \dots, x_n]^\top$ và trả về một giá trị vô hướng (scalar), ta không thể xác định một tốc độ thay đổi chung khi tất cả các biến cùng biến thiên một cách hỗn loạn. Thay vào đó, thực hiện khảo sát tốc độ thay đổi của hàm số theo một biến số duy nhất, trong khi xem tất cả các biến số còn lại như những hằng số cố định.

Đạo hàm riêng (Partial Derivative) của hàm số $f(\mathbf{x})$ theo biến $x_i$ tại điểm $\mathbf{x}_0$ được ký hiệu bằng ký tự toán học $\partial$ (del):
$$\frac{\partial f}{\partial x_i}(\mathbf{x}_0) = \lim_{h \to 0} \frac{f(x_1, \dots, x_i + h, \dots, x_n) - f(x_1, \dots, x_i, \dots, x_n)}{h}$$

Về mặt hình học trong không gian ba chiều, hàm số $f(x, y)$ tạo nên một bề mặt cong (surface). Đạo hàm riêng theo biến $x$ ($\frac{\partial f}{\partial x}$) chính là hệ số góc của đường tiếp tuyến với bề mặt tại điểm đang xét, nhưng đường tiếp tuyến này phải nằm trong mặt phẳng song song với trục hoành $Ox$ (nơi biến $y$ cố định). Tương tự, $\frac{\partial f}{\partial y}$ là hệ số góc của tiếp tuyến nằm trong mặt phẳng song song với trục tung $Oy$.

**Ví dụ**: Xét hàm mục tiêu phi tuyến số thực sau:
    $$f(x, y) = x^3 y^2 + \ln(x) + e^y$$
- Tính đạo hàm riêng theo biến $x$: 
    - Xem $y$ như một hằng số. 
    - Khi đó, đại lượng $e^y$ đóng vai trò là hằng số tự do (đạo hàm bằng 0), và $y^2$ là hệ số nhân đi kèm của $x^3$.
    $$\frac{\partial f}{\partial x} = \frac{\partial}{\partial x}(x^3 y^2) + \frac{\partial}{\partial x}(\ln(x)) + \frac{\partial}{\partial x}(e^y) = 3x^2 y^2 + \frac{1}{x} + 0$$
- Tính đạo hàm riêng theo biến $y$: 
    - Xem $x$ như một hằng số. 
    - Khi đó, đại lượng $\ln(x)$ là hằng số tự do, và $x^3$ là hệ số nhân đi kèm của $y^2$.
    $$\frac{\partial f}{\partial y} = \frac{\partial}{\partial y}(x^3 y^2) + \frac{\partial}{\partial y}(\ln(x)) + \frac{\partial}{\partial y}(e^y) = 2x^3 y + 0 + e^y$$

#### **2.2. Gradient Vector**
Để quản lý các đạo hàm riêng nằm rải rác một cách hệ thống, chúng ta hợp nhất tất cả các thông tin sai phân này vào một cấu trúc toán học duy nhất gọi là Gradient, ký hiệu là $\nabla f(\mathbf{x})$ hoặc $\frac{\partial f}{\partial \mathbf{x}}$.

Về mặt vật lý, Gradient tại một điểm trên trường vô hướng là một véc-tơ mang hai đặc trưng cốt lõi:
- Hướng: Luôn chỉ về hướng mà hàm số có tốc độ tăng trưởng nhanh nhất (steepest ascent).
- Độ lớn: Khẳng định tốc độ tăng trưởng cực đại tại điểm đó.

Trong giải tích ma trận, việc định hình hình học cho các cấu trúc đạo hàm vô cùng phức tạp do tồn tại hai trường phái đối nghịch: Numerator Layout (Bố cục theo tử số / trường phái Jacobian) và Denominator Layout (Bố cục theo mẫu số / trường phái Hessian). Sự bất đồng quy ước này thường dẫn đến các sai số về phép chuyển vị ($\top$) trong quá trình lập trình thuật toán.

Nhằm đảm bảo tính đồng bộ hoàn hảo với quy tắc chuỗi tổng quát (Generalized Chain Rule) và các hệ thống thư viện tính toán tự động (Autograd), chúng ta thống nhất sử dụng Quy ước Bố cục Tử số (Numerator Layout).

Định luật Numerator Layout: Nếu ta lấy đạo hàm của một đại lượng tử số có cấu trúc dạng $M$ theo một đại lượng mẫu số có cấu trúc dạng $N$, thì cấu trúc của ma trận kết quả sẽ giữ nguyên cấu trúc không gian của tử số $M$ làm chiều chủ đạo (chiều hàng) và chuyển vị cấu trúc của mẫu số $N$ làm chiều phụ thuộc (chiều cột).

Áp dụng trực tiếp vào bài toán: Hàm số $f(\mathbf{x})$ là một đại lượng vô hướng (kích thước $1 \times 1$), còn đầu vào $\mathbf{x}$ là một véc-tơ cột thuộc không gian $\mathbb{R}^{n \times 1}$:
$$\mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \\ \vdots \\ x_n \end{bmatrix}$$


**Ví dụ**:

Xét một hàm số hai biến mô tả độ cao của một địa hình đồi núi trong không gian 3 chiều:
    $$f(x, y) = x^2 + 2y^2$$
Hàm số này tạo nên một bề mặt cong có dạng một thung lũng hình parabol.

Yêu cầu: Tìm véc-tơ Gradient của hàm số tại điểm $A(2, 1)$.

Cách giải:
- Bước 1: Tính các đạo hàm riêng thành phần
    - Đầu tiên, ta lấy đạo hàm riêng của $f$ theo biến $x$ (coi $y$ là hằng số):
    $$\frac{\partial f}{\partial x} = \frac{\partial}{\partial x}(x^2) + \frac{\partial}{\partial x}(2y^2) = 2x + 0 = 2x$$
    - Tiếp theo, ta lấy đạo hàm riêng của $f$ theo biến $y$ (coi $x$ là hằng số):
    $$\frac{\partial f}{\partial y} = \frac{\partial}{\partial y}(x^2) + \frac{\partial}{\partial y}(2y^2) = 0 + 4y = 4y$$
- Bước 2: Thành lập véc-tơ Gradient tổng quát
    - Áp dụng Quy ước Bố cục Tử số (Numerator Layout) đã thống nhất ở Phần 2, đạo hàm của hàm vô hướng theo véc-tơ đầu vào sẽ được xếp thành một véc-tơ hàng (Row Vector):
    $$\nabla f(x, y) = \begin{bmatrix} \frac{\partial f}{\partial x} & \frac{\partial f}{\partial y} \end{bmatrix} = \begin{bmatrix} 2x & 4y \end{bmatrix}$$
- Bước 3: Tính giá trị Gradient tại điểm $A(2, 1)$
    - Thay tọa độ $x = 2$ và $y = 1$ vào véc-tơ Gradient tổng quát, ta thu được:
    $$\nabla f(2, 1) = \begin{bmatrix} 2(2) & 4(1) \end{bmatrix} = \begin{bmatrix} 4 & 4 \end{bmatrix}$$

Kết quả: véc-tơ Gradient $\begin{bmatrix} 4 & 4 \end{bmatrix}$ tại điểm $A(2, 1)$
- Về hướng (Direction): Nếu đang đứng tại tọa độ $(2, 1)$ trên sườn đồi, véc-tơ $\begin{bmatrix} 4 & 4 \end{bmatrix}$ (chỉ về hướng chếch góc $45^\circ$ trong mặt phẳng tọa độ) chính là hướng đi lên dốc cao nhất. Nếu bước theo hướng này sẽ leo lên đỉnh đồi với tốc độ nhanh nhất.
- Về độ lớn (Magnitude): Độ dài của véc-tơ Gradient thể hiện độ dốc (tốc độ thay đổi) tại điểm đó. Ta tính độ lớn bằng định lý Pytago:
    $$\|\nabla f(2, 1)\| = \sqrt{4^2 + 4^2} = \sqrt{32} \approx 5.66$$
    - Con số $5.66$ này cho biết: tại điểm $A$, nếu dịch chuyển 1 đơn vị khoảng cách theo hướng dốc nhất, độ cao thực tế trên bề mặt đồi sẽ tăng lên xấp xỉ $5.66$ đơn vị.

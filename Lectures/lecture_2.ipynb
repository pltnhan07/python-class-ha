## **Part 2: The `while` Loop**

#### **1. Basic Syntax & Execution**

```python
counter = 1

while counter <= 3:
    print(f"Processing batch {counter}...")
    counter += 1  

print("All batches processed.")
```

```python
battery_level = 15.0

while battery_level < 20.0:
    print(f"Battery at {battery_level}%. Still charging...")
    battery_level += 2.5

print("Battery has reached minimum operational level (20%).")
```

#### **2. The Infinite Loop Danger**

```python
# The Accidental Infinite Loop
processing_files = 5

while processing_files > 0:
    print(f"Files left: {processing_files}")
```

```python
# The Intentional Infinite Loop
while True:
    user_input = input("Type 'exit' to stop the program: ")
    
    if user_input.lower() == 'exit':
        print("Shutting down...")
        break
    else:
        print(f"You typed: {user_input}. Running again...")
```

#### **3. Loop Control**

##### **`break`**

```python
sensor_temperature = 50
max_safe_temp = 90

while sensor_temperature < 100:
    print(f"Current Temp: {sensor_temperature}C")
    sensor_temperature += 15 
    
    if sensor_temperature >= max_safe_temp:
        print("CRITICAL WARNING: Overheating! Triggering emergency shutdown.")
        break
```

##### **`continue**

```python
countdown = 5

while countdown > 0:
    countdown -= 1
    
    if countdown == 3:
        print("Skipping number 3...")
        continue 
        
    print(f"Counting: {countdown}")
```

#### **4. The `while...else` Clause**
- We can attach an `else` statement to the very end of a `while` loop.
- The `else` block ONLY runs if the loop finishes naturally (when the condition becomes False).
- If the loop is killed by a `break`, the `else` block is skipped.

```python
search_index = 0
max_search = 3
target_found = False

while search_index < max_search:
    print(f"Searching index {search_index}...")
    search_index += 1
else:
    print("Search complete. Target was NOT found in the dataset.")
```

```python
search_index = 0

while search_index < max_search:
    print(f"Searching index {search_index}...")
    if search_index == 1:
        print("Target FOUND!")
        break
    search_index += 1
else:
    print("Search complete. Target was NOT found in the dataset.")
```

## **Part 3: The `for` Loop**

#### **1. Iterating Through Basic Sequences**

- A `for` loop automatically handles the indexing. We do not need to manually create a counter variable (like `i = 0`). 
- Just tell Python: "For every item in this collection, do something."

##### **Interating over a list**

```python
features = ["Age", "Income", "Location", "Education"]

for feature in features:
    print(f"Adding {feature} to the Machine Learning model.")
```

##### **Interating over a string**

```python
dna_sequence = "GATTACA"

for base in dna_sequence:
    print(f"Processing nucleotide: {base}")
```

##### **Interating over a list of tuples**

```python
patient_data = [(180, 85), (165, 60), (170, 72)]

for height, weight in patient_data:
    bmi = weight / ((height/100) ** 2)
    print(f"Patient BMI: {bmi:.1f}")
```

#### **2. Interating Through Dictionaries**

```python
# Given a dictionary
model_config = {"optimizer": "Adam", "learning_rate": 0.01, "epochs": 50}
```

##### **Interating over keys**

```python
for config_name in model_config:
    print(f"Setting configured for: {config_name}") 
```

##### **Interating over values**

```python
for config_value in model_config.values():
    print(f"Value: {config_value}")
```

##### **Interating over both**

```python
for key, value in model_config.items():
    print(f"Hyperparameter {key} is set to {value}.")
```

#### **4. Advanced Interation**

##### **`enumerate()`**

Getting the item AND its index from a list

```python
classes = ["Cat", "Dog", "Bird"]

for index, animal in enumerate(classes):
    print(f"Class {index}: {animal}.")
```

##### **`zip()`**

Iterating multiple lists at the same time

```python
predictions = ["Spam", "Not Spam", "Spam"]
actual_answers = ["Spam", "Spam", "Spam"]
```

```python
for pred, actual in zip(predictions, actual_answers):
    if pred == actual:
        print(f"Correct Prediction.")
    else:
        print(f"Error. Model guessed {pred}, but it was {actual}.")
```

```python
# Combine zip() and enumerate()
for instance, (pred, actual) in enumerate(zip(predictions, actual_answers)):
    if pred == actual:
        print(f"Instance {instance}: Correct Prediction.")
    else:
        print(f"Instance {instance}: Error. Model guessed {pred}, but it was {actual}.")
```

## **Part 4: Comprehensions (List & Dictionary)**

#### **1. List Comprehensions**

List Comprehension là một cách ngắn và gọn gàng để tạo ra các list mới từ các list hiện có bằng cách sử dụng cú pháp đơn giản kết hợp giữa vòng lặp for và điều kiện if.

Syntax: `new_list = [expression for item in iterable if condition]`

**Example**: Given a list of item prices. Apply a 20% discount to each item. Calculate the final amount that the customer needs to pay. 
- Input: [10.0, 25.5, 9.99]
- Output: 36.4

```python
item_prices = [10.0, 25.5, 9.99]

discounted_prices = [price * 0.8 for price in item_prices]
print(discounted_prices)

final_amount = sum(discounted_prices)
print(f"{final_amount:.1f}")
```

##### **Filtering with `if`**
If item price is smaller than 10.0, no discount.

```python
item_prices = [10.0, 25.5, 9.99]

discounted_prices = [price * 0.8 for price in item_prices if price >= 10.0]

final_amount = sum(discounted_prices)
print(f"{final_amount:.1f}")
```

##### **Binary Mapping with `if/else`**

```python
scores = [10, 5, 4, 8, 7, 9]

results = ["pass" if score >=5 else "failed" for score in scores]
print(results)
```

##### **List comprehension with many lists**

```python
list1 = [30, 50, 110, 40, 15, 75]
list2 = [10, 60, 20, 50]

sum_list = [(n1, n2) for n1 in list1 for n2 in list2 if n1 + n2 > 100]

print(sum_list)
```

##### **Nested List Comprehensions**

Work with Matrices

**Example 1**: Flattening a Matrix

Given a "list of lists" (matrix). Your task is extracting all the individual items into one single list (this task is called "flattening").

```python
score_students = [
    [7.5, 8.0, 9.5, 8.0],
    [5.0, 6.4, 7.2, 5.5],
    [8.7, 6.8, 4.3, 9.2],
    [7.7, 9.4, 6.7, 8.8]
]
```

```python
# Method 1: using for loop
scores_1 = []

for student in score_students:
    for score in student:
        scores_1.append(score)
print(scores_1)
```

```python
# Method 2: using list comprehension
# Syntax: [Expression Outer_Loop Inner_Loop]
scores_2 = [score for student in score_students for score in student]
print(scores_2)
```

```python
# Test
assert(scores_1 == scores_2)
```

**Example 2**: Extract only scores that are greater than 8.

```python
over_80 = [score for student in score_students for score in student if score > 8]
print(over_80)
```

**Example 3**: Create a 3x3 matrix filled with zeros.

```python
# Method 1: using for loop
zero_matrix_1 = []

for row in range(3):
    current_row = []
    for cell in range(3):
        current_row.append(0)
    zero_matrix_1.append(current_row)

print(zero_matrix_1)
```

```python
# Method 2: using list comprehension
zero_matrix_2 = [[0 for cell in range(3)] for row in range(3)]
print(zero_matrix_2)
```

#### **2. Dictionary Comprehension**

Dictionary Comprehension là một cách ngắn gọn và mạnh mẽ để tạo dictionary mới từ một iterable (như list, tuple, hay một dictionary khác). Cú pháp của nó tương tự như List Comprehension nhưng kết quả là một dictionary.

**Example 1**: Create a dictionary of numbers up to 6 and their squares.

```python
# Method 1: using for loop
square = {}

for x in range(7):
    square[x] = x**2

print(square)
```

```python
# Method 2: using dictionary comprehension
square = {x: x**2 for x in range(7)}
print(square)
```

**Example 2**: Filter the dictionary of phone book with all phone numbers excepting 111111111.

```python
phone_book = {'John': 123456789, 'Alice': 111111111, 'Bob': 456789123}
```

```python
# Method 1: using for loop
filtered_phone_book_1 = {}

for name, phone in phone_book.items():
    if (phone != 111111111):
        filtered_phone_book_1[name] = phone

print(filtered_phone_book_1)
```

```python
# Method 2: using dictionary comprehension
filtered_phone_book = {k: v for k, v in phone_book.items() if v != 111111111}
print(filtered_phone_book)
```

**Example 3**: Given two lists. Merge them into one dictionary, in which keys and values are elements in list 1 and list 2 respectively.

```python
list1 = ['student 1', 'student 2', 'student 3']
list2 = [96, 80, 74] 
```

```python
# Method 1: using for loop
merge_list_1 = {}

for student, score in zip(list1, list2):
    merge_list_1[student] = score

print(merge_list_1)
```

```python
# Method 2: using dictionary comprehension
merge_list_2 = {student: score for student, score in zip(list1, list2)}
print(merge_list_2)
```

# **OOP**

### **1. Definition**

- OOP là viết tắt của Object-oriented programming (lập trình hướng đối tượng). 
- OOP là một mô hình lập trình dựa trên khái niệm Object (đối tượng).

OOP có 2 thành phần chính:
1. Class (Lớp)
2. Object (Đối tượng)

**Class** (lớp) trong OOP là các template hay cấu trúc để phục vụ cho việc xây dựng nên các Object (đối tượng). Trong đó sẽ bao gồm một tập hợp các Attribute và Methods định nghĩa các đặc tính và hành vi cho các Object.
- Attributes (thuộc tính): định nghĩa các thông tin, đặc điểm cũng như các thuộc tính của Object.
- Methods (phương thức): định nghĩa các hành vi, phương thức cũng như các hành động thường có của Object.

**Object** (đối tượng) là một instance của Class, có thể hiểu là một hiện thực hay ví dụ cụ thể của một Class. 

=> Khi khởi tạo Object, phải truyền đầy đủ thông tin cụ thể mà Class đã định nghĩa.

**Example**:

````python
class Person:
    def __init__(self, name, age, job):
        self.name = name
        self.age = age
        self.job = job
    
    def isTeenager(self):
        return True if self.age < 18 else False

    def work(self):
        return self.name + " is " + self.job
````

````python
# initialize the object p1: an instance of Class Person
p1 = Person("A", 23, "student")
````

````python
# call p1 with its method
p1.isTeenager()
````

````python
p1.work()
````

Class `Person` có:
- Thuộc tính: 3 (`name`, `age`, `job`)
- Phương thức: 2 (`isTeenager()`, `work()`)
- Hàm constructor hay còn gọi là initialize method (phương thức khởi tạo):
```bash
def __init__(self, name, age, job):
        self.name = name
        self.age = age
        self.job = job
```

- `name`, `age`, `job` được gọi là instance variables của class Person.
- `isTeenager()`, `work()` được gọi là instance methods của class Person.

### **2. Constructor**

A constructor is:
- a special method used to create and initialize an object of a class.
- is defined in the class.
- is executed automatically at the time of object creation.

The primary use of a constructor is to declare and initialize instance variables of a class.

For example, when we execute `p1 = Person()`, Python gets to know that `p1` is an object of class `Person` and calls the constructor of that class to create an object.

Syntax:
```bash
def __init__(self):
```

Where,
- `def`: The keyword is used to define function.
- `__init__()` Method: It is a reserved method. This method gets called as soon as an object of a class is instantiated.
- `self`: The first argument self refers to the current object. It binds the instance to the `__init__()` method.

There are 3 types of constructors in Python.
1. Default Constructor
2. Non-parameterized Constructor
3. Parameterized Constructor


##### **2.1. Default Constructor**

- When we do not include the constructor in the class, Python will add a default constructor.
- The default constructor does not perform any task but initializes the objects. It is an empty constructor without a body.

**Example**:

````python
class Person:
    def work(self):
        print("They are teachers.")
````

````python
p1 = Person()
p1.work()
````

##### **2.2. Non-parameteried Constructor**

- a constructor without any arguments
- is used to initialize each object with default values
- doesn’t accept the arguments during object creation (because it initializes every object with the same set of values.)

**Example**:

````python
class Person:
    def __init__(self):
        self.name = "A"
        self.age = 23
        self.job = "student"
    
    def isTeenager(self):
        return True if self.age < 18 else False
    
    def work(self):
        return self.name + " is a/an " + self.job
````

````python
p1 = Person()
p1.isTeenager()
````

````python
p1.work()
````

##### **2.3. Parameteried Constructor**

- constructor with defined parameters or arguments 
- can pass different values to each object at the time of creation using a parameterized constructor

````python
class Person:
    def __init__(self, name, age, job):
        self.name = name
        self.age = age
        self.job = job
    
    def isTeenager(self):
        return True if self.age < 18 else False

    def work(self):
        return self.name + " is " + self.job
````

````python
p1 = Person("A", 23, "student")
print(p1.work())
print(p1.isTeenager())
````

##### **2.4. Constructor with Default Values**

The default value will be used if we do not pass arguments to the constructor at the time of object creation.

````python
class Person:
    def __init__(self, name="A", age=23, job="student"):
        self.name = name
        self.age = age
        self.job = job
    
    def isTeenager(self):
        return True if self.age < 18 else False

    def work(self):
        return self.name + " is " + self.job
````

### **3. Notes**

- `self` is a keyword. Use it to access the instance variables and methods.
- Declare only one constructor for a class.


````python
p1 = Person()
p1.work()
````

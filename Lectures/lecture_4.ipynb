## **Exception Handling**

Có 2 kiểu lỗi có sẵn trong python:
- Syntax Error: Có tên gọi khác là Parsing Errors (lỗi phân tích cú pháp), đây là lỗi cơ bản nhất, xảy ra khi Python parser, trình phân tích cú pháp của Python, không thể hiểu được một dòng code cụ thể nào đó.
-Exception Error: Là những lỗi phát sinh trong quá trình chương trình thực thi

Exception trong python là:
- Một sự kiện xảy ra trong quá trình thực thi một chương trình. Khi nó xảy ra, ứng dụng hoặc server có thể bị crash, khi đó Python sẽ tạo ra một exception để xử lý cho vấn đề đó. 
- Bất kỳ điều kiện bất thường nào trong chương trình làm gián đoạn luồng thực thi chương trình đó thì đều được coi là ngoại lệ. Bất kể khi nào xuất hiện một ngoại lệ, chương trình sẽ ngừng thực thi và chuyển qua quá trình gọi & in ra lỗi cho đến khi nó được xử lý.

#### **2. `try/except` block**

Sử dụng khối lệnh `try/except` để xử lý exception.

Cấu trúc cơ bản:
```bash
try:
    # đoạn code có thể gây lỗi
except SomeException as e:
    # xử lý lỗi
```

Ghi lại trường hợp lỗi (dưới dạng e):
- Khi bắt một lỗi cụ thể, có thể sử dụng cú pháp `except ExceptionType as e`. Biến `e` ghi lại chính xác thông báo hệ thống do Python tạo ra, cho phép chúng ta ghi lại lỗi toán học chính xác.

Quy tắc thứ tự ưu tiên:
- Nếu xếp chồng nhiều khối `except`, Python sẽ đánh giá chúng từ trên xuống dưới.

=> Phải luôn sắp xếp chúng từ lỗi cụ thể nhất đến lỗi tổng quát nhất.

#### **3. `finally`/`else` clauses**

Mệnh đề `else`: 
- Mã bên trong khối `else` chỉ được thực thi nếu khối `try` hoàn thành hoàn hảo mà không gây ra bất kỳ ngoại lệ nào. Đây là vị trí tối ưu để lưu các nghiệm đã tính toán vào cơ sở dữ liệu hoặc thêm chúng vào một danh sách, đảm bảo rằng chỉ những số hợp lệ mới được lưu trữ.

Mệnh đề `finally`: 
- Mã bên trong khối `finally` được thực thi vô điều kiện, bất kể phép tính có thành công hay thất bại. Trong xử lý số học quy mô lớn, `finally` được sử dụng để đóng các luồng dữ liệu, giải phóng bộ nhớ hệ thống hoặc tăng bộ đếm toàn cục cho biết một phương trình đã được xử lý.

##### **Example 1**

````python
try:
    numerator = float(input("Enter a numerator: "))
    denominator = float(input("Enter a denominator: "))
    root = numerator / denominator
except ValueError as e:
    print(f"Data Error: {e}")
except ZeroDivisionError:
    print("Mathematical Error: The denominator cannot be zero.")
else:
    print(root)
finally:
    print("Done")
````

````python
# Example 2
import math

def quadratic_solver(equation_data: dict) -> float:
    try:
        a = equation_data["a"]
        b = equation_data["b"]
        c = equation_data["c"]
        
        discriminant = (b ** 2) - (4 * a * c)
        root = (-b + math.sqrt(discriminant)) / (2 * a)
    except KeyError as e:
        print(f"Data Error: Missing the {e} coefficient.")
        return None
    except TypeError as e:
        print(f"Calculation failed: {e}")
        return None
    except ValueError as e:
        print(f"Discriminant: {discriminant} < 0. The equation has no solution!")
        return None
    except ZeroDivisionError as e:
        print(f"a = {a}. The equation is first-degree.")
        if b != 0:
            linear_root = -c/b
            return linear_root
        else:
            return None
    else:
        if (discriminant > 0):
            root1 = (-b + math.sqrt(discriminant)) / (2 * a)
            root2 = (-b - math.sqrt(discriminant)) / (2 * a)
            return root1, root2
        elif (discriminant == 0):
            return root
    finally:
        print("Done")
````

````python
import math
from typing import Union

def quadratic_solver(equation_data: dict) -> Union[tuple[float, float], float, None]:
    
    # 1. THE UNPREDICTABLE DATA (Handled with try/except)
    try:
        # We attempt to extract and immediately convert to floats.
        # This catches KeyErrors (missing data) and TypeErrors/ValueErrors (bad data types)
        a = float(equation_data["a"])
        b = float(equation_data["b"])
        c = float(equation_data["c"])
    except KeyError as e:
        print(f"Data Error: Missing coefficient {e}.")
        return None
    except (TypeError, ValueError):
        print("Data Error: All coefficients must be convertible to numbers.")
        return None

    # 2. THE PREDICTABLE MATHEMATICS (Handled with Look-Before-You-Leap logic)
    
    # Check linear boundary BEFORE division by zero occurs
    if a == 0.0:
        print(f"a = 0. The equation is first-degree.")
        return -c / b if b != 0 else None
        
    discriminant = (b ** 2) - (4 * a * c)
    
    # Check complex boundary BEFORE math.sqrt() crashes
    if discriminant < 0:
        print(f"Discriminant: {discriminant} < 0. No real solutions.")
        return None
        
    elif discriminant == 0:
        return -b / (2 * a)
        
    else:
        # We only run the heavy math function if we proved it is safe
        sqrt_val = math.sqrt(discriminant)
        return (-b + sqrt_val) / (2 * a), (-b - sqrt_val) / (2 * a)
````

````python
dataset_4 = {"a": 0.6, "b": 5.0, "c": 6.0}
quadratic_solver(dataset_4)
````

#### **4. `raise` keyword**

**Example**: Viết hàm thực thi lệnh rút tiền khỏi tài khoản ngân hàng `withdraw_money()`. 
- Input: 
    - `account_balance`: số dư tài khoản
    - `withdraw_amount`: số tiền muốn rút
- Output:
    - `left_account_balance`: số dư tài khoản sau khi thực hiện lệnh

Điều kiện để thực thi được lệnh rút tiền: số dư tài khoản phải lớn hơn hoặc bằng số tiền muốn rút.

````python
def withdraw_money(account_balance: float, withdraw_amount: float) -> float:
    if (withdraw_amount > account_balance):
        raise ValueError("Not enough money to withdraw!")
    
    left_account_balance = account_balance - withdraw_amount

    return left_account_balance
````

````python
account_balance = 500.0
withdraw_amount = 122.5
withdraw_money(account_balance, withdraw_amount)
````

````python
account_balance = 500.0
withdraw_amount = 522.5
withdraw_money(account_balance, withdraw_amount)
````

#### **5. Assertions**

- Câu lệnh dùng để kiểm tra một điều kiện có đúng hay không trong quá trình thực thi chương trình. 
- Nếu điều kiện sai, Python sẽ ngay lập tức trả về một ngoại lệ AssertionError, giúp phát hiện và sửa lỗi sớm hơn.

Syntax: 
```bash
assert condition, message
```

Trong đó:
- `condition`: Một biểu thức điều kiện phải đúng để chương trình tiếp tục chạy.
- `message` (optional): Thông báo lỗi hiển thị khi assertion thất bại, giúp xác định nguyên nhân lỗi dễ dàng hơn.

````python
score = 5.2
assert 0.0 <= score <= 10.0
````

````python
score = 15.2
assert 0.0 <= score <= 10.0
````

````python
score = 15.2
assert 0.0 <= score <= 10.0, "score is out of range"
````

**Example 1**: Nhập danh sách điểm của 10 học sinh. Báo lỗi nếu điểm nhập vào không nằm trong khoảng [0.0,  10.0]. In danh sách điểm.

````python
scores = [float(input("Nhập điểm: ")) for i in range(10)]
assert (0.0 <= score <= 10.0 for score in scores)
print(scores)
````

````python
scores = [float(input("Nhập điểm: ")) for i in range(10)]
assert all(0.0 <= score <= 10.0 for score in scores), "score is out of range"
print(scores)
````

##### **AssertionError Handling**

- AssertionError là ngoại lệ được đưa ra khi một điều kiện kiểm tra bằng assert không được thỏa mãn. 
- Mặc định, nếu lỗi này xảy ra mà không được xử lý, chương trình sẽ dừng ngay lập tức và hiển thị traceback. 
- Tuy nhiên, có thể bắt và xử lý AssertionError bằng cách sử dụng khối try-except. Điều này giúp chương trình tiếp tục hoạt động mà không bị gián đoạn.

````python
try:
    scores = [float(input("Nhập điểm: ")) for i in range(10)]
    assert all(0.0 <= score <= 10.0 for score in scores), f"score {score} is out of range"
    print(scores)
except AssertionError as e:
    print(e)
````

##### **Assertions vs Exceptions**

- Exceptions:
    - Xử lý các yếu tố bên ngoài không thể đoán trước. 
    - Bảo vệ chương trình khỏi các lỗi nhập liệu do người dùng gây ra.
- Assertions: 
    - Xử lý các lỗi nội bộ. (VD: kiểm tra các tính toán có chính xác?)

````python
# Case 1: raise an exception
def withdraw_money(account_balance: float, withdraw_amount: float) -> float:
    if (withdraw_amount > account_balance):
        raise ValueError("Not enough money to withdraw!")
    
    left_account_balance = account_balance - withdraw_amount

    return left_account_balance
````

````python
account_balance = 500.0
withdraw_amount = 122.5
withdraw_money(account_balance, withdraw_amount)
````

````python
account_balance = 500.0
withdraw_amount = 522.5
withdraw_money(account_balance, withdraw_amount)
````

````python
# Case 2: assert
def withdraw_money(account_balance: float, withdraw_amount: float) -> float:
    assert (withdraw_amount <= account_balance), "Not enough money to withdraw!"
    
    left_account_balance = account_balance - withdraw_amount
    
    return left_account_balance
````

````python
account_balance = 500.0
withdraw_amount = 122.5
withdraw_money(account_balance, withdraw_amount)
````

````python
account_balance = 500.0
withdraw_amount = 522.5
withdraw_money(account_balance, withdraw_amount)
````

# **Numerical Computing with Numpy**
NumPy (Numerical Python) is a module which was created allow efficient numerical calculations on multi-dimensional arrays of numbers from within Python. It is derived from the merger of two earlier modules named Numeric and Numarray. The actual work is done by calls to routines written in the Fortran and C languages.

- NumPy defines a new data type called the `ndarray` (n-dimensional array).
- The elements of ndarrays can be integers, floats, and complex numbers of various sizes. 
- When the NumPy package is loaded, ndarrays become as much a part of the Python language as standard Python data types such as lists and dictionaries.
- A NumPy array can only hold one single data type at a time.

### **4. Array data types**


##### ́**4.1. Integer**
- Signed Integer: The number trailing the type indicates the bit-width. The more bits you use, the larger the numbers you can store, but the more RAM you consume.
    - int8: Ranges from -128 to 127.
    - int16: Ranges from -32,768 to 32,767.
    - int32: Ranges from $-2.14 \times 10^9$ to $2.14 \times 10^9$.
    - int64: Ranges from $-9 \times 10^{18}$ to $9 \times 10^{18}$.
- Unsigned Integer: By dropping the negative numbers, unsigned integers shift their entire mathematical range into the positive zone.
    - uint8: Ranges from 0 to 255.
    - uint16: Ranges from 0 to 65,535.
    - uint32: Ranges from 0 to $4.29 \times 10^9$.
    - uint64: Ranges from 0 to $18.4 \times 10^18$.

````python
a = np.array([1, 2], dtype = np.int8)
a
````

````python
b = np.array([-11, 2], dtype = np.uint8)
b
````

- "0": Targets Row 0

##### **5.4. Reshaping arrays**
`reshape` method allows the number of dimensions and the size of each dimension to be changed as long as the total number of array elements remains the same.

````python
a = np.array([[1, 2], [3, 4]])
a
````

````python
a.reshape(4)
````

- The numbers in array `a` are read row-by-row 
- And unpacked into a single horizontal row.

````python
a.reshape(4,1)
````

````python
a.reshape(2,2,1)
````

- 2 (First Dimension): Creates 2 separate matrices.
- 2 (Second Dimension): Each of those matrices contains 2 rows.
- 1 (Third Dimension): Each row contains exactly 1 element.

````python
a
````

````python
b = a.reshape(4)
b
````

````python
a[0,1] = 13
a[1,0] = 45
b
````

As with slicing, reshaping doesn’t normally create a new array, just a new view of the original array.

##### **5.5. Array looping**

**1D array**

````python
a = np.arange(0,5)
a
````

````python
for item in a:
    print(item)
````

**2D array**

````python
a = np.array([[1, 2], [3, 4]])
a
````

````python
for row in a:
    print(row)
````

````python
for row in a:
    for item in row:
        print(item)
````

**`flat` attribute**

````python
a = np.array([[1, 2], [3, 4]])
a
````

````python
for item in a.flat:
    print(item)
````

**`np.ndenumerate()` method**

````python
a = np.array([[1, 2], [3, 4]])
a
````

````python
for index, item in np.ndenumerate(a):
    print(f"{index} has value {item}")
````

**`np.nditer()` method**

````python
a = np.array([[1, 2], [3, 4]])
a
````

````python
for item in np.nditer(a):
    print(item)
````

````python
with np.nditer(a, op_flags=['readwrite']) as iter:
    for item in iter:
        item[...] = item*10

a
````

````python
a = np.array([[1, 2], [3, 4]])
a
````

````python
with np.nditer(a, op_flags=['readwrite']) as iter:
    for item in iter:
        item[...] = item*10 if item%2==0 else item

a
````

##### **Broadcasting**

````python
a = np.array([[1, 2], [3, 4]])
a
````

````python
e = np.array([10, 20, 30])
e
````

````python
a + e
````

The shape of array a is (2,2),while the shape of array e is (3,), and they therefore cannot be added together.

````python
a.shape
````

````python
e.shape
````

````python
b = np.array([[5,10]])
b.shape
````

````python
a + b
````

In this case array b, which has one row and two columns, is (conceptually) extended along axis 0, repeating values as necessary, to create an array with two rows array([[5,10],[5,10]]) and then added to a. This extension process is called broadcasting.

````python
c = np.array([[5], [10]])
c.shape
````

````python
a + c
````

In this case array c is extended along axis 1, to create an array with two rows array([[5,5],[10,10]]), which is then added to a. Notice that the result here is different than that in the previous case, since the broadcast axis is different.

````python
b + c
````

That is a case of double broadcasting.

Để hai mảng có thể thực hiện broadcasting với nhau, NumPy sẽ so sánh shape của chúng từ phải qua trái (từ chiều cuối cùng lên chiều đầu tiên). Hai chiều được coi là tương thích khi và chỉ khi:
- Chúng bằng nhau.
- Hoặc một trong hai chiều có kích thước bằng 1.

````python
d = np.array([5, 10])
d
````

````python
d.shape
````

````python
a + d
````

````python
d1 = d.reshape(1,2)
d1
````

````python
a + d1
````

````python
d2 = d.reshape(2,1)
d2
````

````python
a + d2
````

#### **2. Parameters vs. Arguments**

These terms are often used interchangeably, but technically:
- Parameter: The placeholder variable in the function definition.
- Argument: The actual data (value) you pass in when you call it.

Consider the algebraic function of two variables:

$$f(x, y) = x^2 + y^2$$

```python
# 'x' and 'y' are the PARAMETERS
def evaluate_z(x, y):
    z = (x ** 2) + (y ** 2)
    print(f"f({x}, {y}) = {z}")
```

```python
# 3 and 4 are the ARGUMENTS
evaluate_z(3, 4) 
```

#### **3. The `return` Statement**

Two points: $I_1(x_1, y_1)$, $I_2(x_2, y_2)$

Consider the Euclidean distance formula:

$$d(I_1, I_2) = \sqrt{(x_1 - x_2)^2 + (y_1 - y_2)^2}$$

```python
import math
def get_distance_from_origin(x, y):
    distance = math.sqrt(x**2 + y**2)
    return distance
```

```python
i1_distance = get_distance_from_origin(0.5, 0.5)
print(i1_distance)
is_inside_unit_circle = i1_distance <= 1.0
print(is_inside_unit_circle)

i2_distance = get_distance_from_origin(0.7, 1.2)
print(i2_distance)
is_inside_unit_circle = i2_distance <= 1.0
print(is_inside_unit_circle)
```

```python
import math
def get_distance(i1, i2):
    distance = math.sqrt((i1[0] - i2[0])**2 + (i1[1] - i2[1])**2)
    return distance
```

```python
i1 = (0.5, 0.5)
i2 = (0.7, 1.2)
d = get_distance(i1, i2)
print(d)
```

#### **4. Returning Multiple Values**

```python
def solve_quadratic(a, b, c):
    discriminant = (b ** 2) - (4 * a * c)
    if discriminant > 0:
        root1 = (-b + math.sqrt(discriminant)) / (2 * a)
        root2 = (-b - math.sqrt(discriminant)) / (2 * a)
        
        return root1, root2
    
    elif discriminant == 0:
        root = -b / (2 * a)
        
        return root
    
    else:
        return None
```

```python
equation1 = solve_quadratic(1, -5, 6)
print(equation1)
```

```python
equation2 = solve_quadratic(1, -5, 8)
print(equation2)
```

```python
equation3 = solve_quadratic(1, 2, 1)
print(equation3)
```

```python
import math

def compute_l1_distance(p1, p2):
    diff = abs(p1 - p2)
    return diff

def compute_l2_distance(p1, p2):
    diff = (p1 - p2) ** 2
    return math.sqrt(diff)
```

```python
GRAVITATIONAL_CONSTANT = 6.6743e-11 

def calculate_gravitational_force(mass1, mass2, distance):
    force = GRAVITATIONAL_CONSTANT * (mass1 * mass2) / (distance ** 2)
    return force
```

#### **3. The `global` keyword**

- By default, a function can read a global variable, but it cannot overwrite it. 
- If you try to change a global variable within a function, Python will throw an error or shadow it locally. 
- To modify a global state from within a function, you must use the `global` keyword.

```python
i = 0

def square(x):
    global i
    i += 1  
    
    return (x - i)
```

```python
print(square(2.0))
print(square(2.1))
print(square(1.9))

print(f"i = {i}")
```

#### **4. The `nonlocal` keyword**
- The Enclosing Scope (also known as non-local scope) occurs when you nest a function inside another function.
- The inner function automatically inherits read-access to the variables of the outer function.
- If the inner function needs to modify a variable belonging to the outer function, it must declare that variable as `nonlocal`.

```python
def outer():
    message = "Hello"
    print("Inside outer:", message)
    
    def inner():
        nonlocal message 
        message = "Hello, Python!"
        print("Inside inner:", message)
    
    inner()
    print("Inside outer:", message)

outer()
```

#### **4. Arbitrary Arguments**

Allow functions to accept multiple values. This is done using two special symbols:
- `*args` collects extra positional arguments as a tuple.
- `**kwargs` collects extra keyword arguments as a dictionary.

```python
def non_key(*args):
    print("Non-Keyword Arguments (*args):")
    for arg in args:
        print(arg)
```

```python
non_key('Hey', 'Welcome', 'An')
```

```python
def key(**kwargs):
    print("Keyword Arguments (**kwargs):")
    for key, value in kwargs.items():
        print(f"{key} == {value}")
```

```python
key(first='Hey', mid='Welcome', last='An')
```

```python
key('Hey', 'Welcome', 'An')
```

```python
key(f='Hey', m='Welcome', l='An')
```

## **Part 4: Lambda Function**

Lambda functions:
- Similar to user-defined functions but without a name.
- Referred to anonymous functions.
- Efficient whenever creating a function that only contains simple expressions:
    - Expressions with a single line of a statement. 
    - Functions that will be used once.

Syntax: `lambda argument(s) : expression`
- `lambda`: keyword
- `argument(s)`: is a placeholder, that is a variable that will be used to hold the value and pass into the function expression.
- `expression` is the code you want to execute in the lambda function.

#### **1. Basic Lambda Function**

##### **Single variable**

```python
# User-defined function
def f(x):
  return x * 2

f(3)
```

```python
# Lambda function
f = lambda x: x * 2
f(3)
```

##### **Multi-variable**

```python
import math

calculate_cylinder_volume = lambda r, h: math.pi * (r ** 2) * h

volume = calculate_cylinder_volume(3.0, 10.0)
print(f"Cylinder Volume: {volume:.4f}")
```

#### **2. Lambda Function with `sort()`**

Suppose we have a list of 2D points $(x, y)$, and we want to sort them based on their Manhattan Distance from the origin $(0,0)$. 

The formula is:
$$d = |x| + |y|$$

```python
coordinates = [(4, -2), (1, 1), (-3, 5), (0, 3)]
```

```python
sorted_by_manhattan = sorted(coordinates, key=lambda point: abs(point[0]) + abs(point[1]))
print(sorted_by_manhattan)
```

#### **3. Lambda Functions with Functional Operators**

##### **`map()`**

The map() function scales operations across an entire sequence. Let's pass a list of raw inputs through the mathematical Sigmoid Activation Function, commonly used to compress values into a probability space between $0$ and $1$:
$$S(x) = \frac{1}{1 + e^{-x}}$$

```python
raw_inputs = [-2.0, -1.0, 0.0, 1.0, 2.0]
```

```python
import math
sigmoid_probabilities = list(map(lambda x: 1 / (1 + math.exp(-x)), raw_inputs))
```

```python
print(sigmoid_probabilities)
```

```python
formatted_output = [f"{p:.4f}" for p in sigmoid_probabilities]
print(formatted_output)
```

##### **`filter()`**

The filter() function tests elements against a conditional boolean expression and drops elements that return False. Let's isolate points from a numeric domain that satisfy the polynomial inequality:
$$x^2 - 3x > 0$$

```python
samples = [-2, -1, 0, 1, 2, 3, 4, 5]
```

```python
valid_samples = list(filter(lambda x: (x ** 2) - (3 * x) > 0, samples))
print(valid_samples)
```

## **Part 5: Type Hinting**

- Statically typed languages: khi khai báo hàm và các parameter của nó thì phải biết được kiểu dữ liệu tại thời điểm biên dịch. Điều này có nghĩa là khi lập trình, phải khai báo kiểu của biến một cách tường minh.

- Dynamically typed languages: kiểu dữ liệu của biến được liên kết với giá trị lúc run-time. Điều này có nghĩa là khi lập trình, không cần khai báo kiểu của một biến. Với một biến được khai báo, có thể tùy ý gán gía trị bất kì cho nó mà không cần quan tâm tới kiểu của nó là gì. Nó sẽ tự động ép kiểu khi nhận được giá trị.

=> Python thuộc loại Dynamically typed languages.

=> Vấn đề: Nhiều trường hợp kiểu của biến sẽ không được tường minh và gây khó khăn trong quá trình debug.

=> Tính năng Type Hints: cung cấp một cú pháp tiêu chuẩn cho chú thích kiểu của dữ liệu đầu vào và đầu ra. 

=> Không làm thay đổi bản chất của Dynamically typed languages.

#### **3. Optional Types and Collections**

Use Optional for values that can be None, and define types for collections like lists and tuples.

```python
from typing import Optional, List, Tuple
```

```python
def get_user(id: int) -> Optional[str]:
    return None if id == 0 else "User"

get_user(5)
```

```python
def sum_list(num: List[int]) -> int:
    return sum(num)

sum_list([2, 3, 5])
```

```python
def get_name_and_age() -> Tuple[str, int]:
    return ("Abc", 25)

get_name_and_age()
```

#### **4. Example Without Type Hints**

```python
# Without Type Hints
def factorial(i):
    if i < 0:
        return None
    if i == 0:
        return 1
    return i * factorial(i - 1)
```

```python
print(factorial(4))
```

```python
print(factorial("4"))
```

```python
print(factorial(5.01))
```

```python
# Add Type Hints
def factorial(i: int) -> int:
    if not isinstance(i, int):  
        return None
    if i < 0:
        return None
    if i == 0:
        return 1
    return i * factorial(i - 1)
```

```python
print(factorial(4))
```

```python
print(factorial("4"))
```

```python
print(factorial(5.01))
```

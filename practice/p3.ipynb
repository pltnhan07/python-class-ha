#### **Q1**:

A lucky number is defined using a special elimination process:
   1. Start with natural numbers: 1, 2, 3, 4, 5, 6, ...
   2. In the 1st step, remove every 2nd number.
   3. In the 2nd step, remove every 3rd remaining number.
   4. In the 3rd step, remove every 4th remaining number, and so on...

This continues indefinitely. Given an integer n, determine if it remains after all steps.

Return 1 if n is a lucky number, otherwise return 0. (1 ≤  n ≤  10^5)

Example:

Input: n = 5

Output: 0


Input: n = 19

Output: 1

#### **Q2**:

Given a number n, find the value of n raised to the power of its own reverse. (1 ≤ n ≤ 10)

Note: The result will always fit into a 32-bit signed integer.

Examples:

Input: n = 2

Output: 4


Input: n = 10

Output: 10

#### **Q3**:

You are given n disks placed on a starting rod (from), with the smallest disk on top and the largest at the bottom. There are three rods: the starting rod(from), the target rod (to), and an auxiliary rod (aux).

You have to calculate the minimum number of moves required to transfer all n disks from the starting rod to the target rod, following these rules:
      1. Only one disk can be moved at a time.
      2. A disk can only be placed on top of a larger disk or on an empty rod.

Return the minimum number of moves needed to complete the task.

Examples:

Input: n = 2

Output: 3


Input: n = 3

Output: 7


Input: n = 0

Output: 0

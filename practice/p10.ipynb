Hiện thực hóa các phép toán đại số tuyến tính cơ bản trên không gian không gian vector $\mathbb{R}^n$ và không gian ma trận $\mathbb{R}^{m \times n}$ thông qua mảng đa chiều của NumPy.  
- Chuẩn $L_2$ (Độ dài vector): Ký hiệu là $||v||_2$, được tính bằng căn bậc hai của tổng bình phương các phần tử: $||v||=\sqrt{v_{1}^{2}+v_{2}^{2}+...+v_{n}^{2}}$.  
- Tích vô hướng (Dot Product): Biểu thị hình chiếu hình học giữa hai vector: $v\cdot u=v_{1}*u_{1}+v_{2}*u_{2}+...+v_{n}*u_{n}$.  
- Tích ma trận - vector: Biến đổi tuyến tính một vector thông qua ma trận $c = Av$.  
- Tích ma trận - ma trận: Tổng hợp hai phép biến đổi tuyến tính $C = AB$.  

Yêu cầu: Viết các hàm `compute_vector_length`, `compute_dot_product`, `matrix_multi_vector`, `matrix_multi_matrix` chỉ sử dụng thư viện numpy.

### Q2: Đo lường Khoảng cách Không gian (Cosine Similarity)
Định lượng mức độ tương đồng giữa hai vector biểu diễn dữ liệu trong không gian đa chiều.  
- Độ tương đồng Cosine đánh giá góc giữa hai vector, bỏ qua độ lớn của chúng. Công thức chuẩn:
    $$cosine(x,y)=\frac{x\cdot y}{||x||||y||}=\frac{\Sigma_{1}^{n}x_{i}y_{i}}{\sqrt{\Sigma_{1}^{n}x_{i}^{2}}\sqrt{\Sigma_{1}^{n}y_{i}^{2}}}$$
- Giá trị trả về nằm trong khoảng $[-1, 1]$, trong đó $1$ thể hiện sự tương đồng tuyệt đối về hướng.  

Yêu cầu: Xây dựng hàm `compute_cosine(v1, v2)`. Tính Cosine Similarity cho hai vector: $x=[1, 2, 3, 4]$ và $y=[1, 0, 3, 0]$
